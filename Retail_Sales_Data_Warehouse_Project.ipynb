```python
# Retail Sales Data Warehouse Project
# Step 1: Import required libraries

import pandas as pd
import sqlite3
from datetime import datetime

print("Data Warehouse Project Started Successfully!")
```

```python
# Step 2: Create Source Sales Data

sales_data = {
    'Transaction_ID': ['T001', 'T002', 'T003', 'T004', 'T005',
                       'T006', 'T007', 'T008', 'T009', 'T010'],
    'Date': ['2026-07-01', '2026-07-02', '2026-07-03', '2026-07-04', '2026-07-05',
             '2026-07-06', '2026-07-07', '2026-07-08', '2026-07-09', '2026-07-10'],
    'Customer_ID': ['C001', 'C002', 'C003', 'C001', 'C004',
                    'C005', 'C002', 'C003', 'C004', 'C005'],
    'Customer_Name': ['Kamal', 'Nimali', 'Saman', 'Kamal', 'Amali',
                      'Nuwan', 'Nimali', 'Saman', 'Amali', 'Nuwan'],
    'City': ['Colombo', 'Kandy', 'Galle', 'Colombo', 'Kurunegala',
             'Colombo', 'Kandy', 'Galle', 'Kurunegala', 'Colombo'],
    'Product_ID': ['P001', 'P002', 'P003', 'P002', 'P001',
                   'P004', 'P003', 'P001', 'P004', 'P002'],
    'Product_Name': ['Laptop', 'Mouse', 'Keyboard', 'Mouse', 'Laptop',
                     'Monitor', 'Keyboard', 'Laptop', 'Monitor', 'Mouse'],
    'Category': ['Electronics', 'Accessories', 'Accessories', 'Accessories',
                 'Electronics', 'Electronics', 'Accessories', 'Electronics',
                 'Electronics', 'Accessories'],
    'Store_ID': ['S001', 'S001', 'S002', 'S001', 'S002',
                 'S001', 'S002', 'S002', 'S001', 'S002'],
    'Quantity': [1, 2, 1, 3, 1, 1, 2, 1, 1, 4],
    'Unit_Price': [250000, 3500, 7500, 3500, 250000,
                   65000, 7500, 250000, 65000, 3500],
    'Discount': [10000, 500, 0, 500, 15000, 5000, 1000, 10000, 3000, 1000]
}

source_sales = pd.DataFrame(sales_data)

source_sales
```

```python
# Step 3: Create Staging Layer

staging_sales = source_sales.copy()

# Convert Date column to proper date format
staging_sales['Date'] = pd.to_datetime(staging_sales['Date'])

# Remove duplicate records
staging_sales = staging_sales.drop_duplicates()

# Check for missing values
print("Missing Values:")
print(staging_sales.isnull().sum())

print("\nStaging Data:")
display(staging_sales)
```

```python
# Step 4: Transform Data and Calculate Measures

# Calculate Gross Sales
staging_sales['Gross_Sales'] = (
    staging_sales['Quantity'] * staging_sales['Unit_Price']
)

# Calculate Net Sales after discount
staging_sales['Net_Sales'] = (
    staging_sales['Gross_Sales'] - staging_sales['Discount']
)

print("Transformation Completed Successfully!")

display(
    staging_sales[
        ['Transaction_ID', 'Product_Name', 'Quantity',
         'Unit_Price', 'Discount', 'Gross_Sales', 'Net_Sales']
    ]
)
```

```python
# Step 5: Create Dimension Tables with Surrogate Keys

# Customer Dimension
dim_customer = staging_sales[
    ['Customer_ID', 'Customer_Name', 'City']
].drop_duplicates().reset_index(drop=True)

dim_customer.insert(0, 'Customer_SK', range(1, len(dim_customer) + 1))


# Product Dimension
dim_product = staging_sales[
    ['Product_ID', 'Product_Name', 'Category']
].drop_duplicates().reset_index(drop=True)

dim_product.insert(0, 'Product_SK', range(1, len(dim_product) + 1))


# Store Dimension
dim_store = pd.DataFrame({
    'Store_ID': ['S001', 'S002'],
    'Store_Name': ['Colombo Store', 'Kandy Store']
})

dim_store.insert(0, 'Store_SK', range(1, len(dim_store) + 1))


# Date Dimension
dim_date = pd.DataFrame({
    'Date': staging_sales['Date'].drop_duplicates().sort_values()
}).reset_index(drop=True)

dim_date['Day'] = dim_date['Date'].dt.day
dim_date['Month'] = dim_date['Date'].dt.month
dim_date['Year'] = dim_date['Date'].dt.year
dim_date.insert(0, 'Date_SK', range(1, len(dim_date) + 1))


print("CUSTOMER DIMENSION")
display(dim_customer)

print("PRODUCT DIMENSION")
display(dim_product)

print("STORE DIMENSION")
display(dim_store)

print("DATE DIMENSION")
display(dim_date)
```

```python
# Step 6: Create Fact Sales Table

fact_sales = staging_sales.copy()

# Add Customer_SK
fact_sales = fact_sales.merge(
    dim_customer[['Customer_SK', 'Customer_ID', 'Customer_Name', 'City']],
    on=['Customer_ID', 'Customer_Name', 'City'],
    how='left'
)

# Add Product_SK
fact_sales = fact_sales.merge(
    dim_product[['Product_SK', 'Product_ID', 'Product_Name', 'Category']],
    on=['Product_ID', 'Product_Name', 'Category'],
    how='left'
)

# Add Store_SK
fact_sales = fact_sales.merge(
    dim_store[['Store_SK', 'Store_ID']],
    on='Store_ID',
    how='left'
)

# Add Date_SK
fact_sales = fact_sales.merge(
    dim_date[['Date_SK', 'Date']],
    on='Date',
    how='left'
)

# Keep only Fact Table columns
fact_sales = fact_sales[
    [
        'Transaction_ID',
        'Date_SK',
        'Customer_SK',
        'Product_SK',
        'Store_SK',
        'Quantity',
        'Unit_Price',
        'Discount',
        'Gross_Sales',
        'Net_Sales'
    ]
]

print("FACT SALES TABLE")
display(fact_sales)
```

```python
# Step 7: Create SQLite Data Warehouse Database

# Create database connection
conn = sqlite3.connect('retail_data_warehouse.db')

# Load Dimension Tables
dim_customer.to_sql('Dim_Customer', conn, if_exists='replace', index=False)
dim_product.to_sql('Dim_Product', conn, if_exists='replace', index=False)
dim_store.to_sql('Dim_Store', conn, if_exists='replace', index=False)
dim_date.to_sql('Dim_Date', conn, if_exists='replace', index=False)

# Load Fact Table
fact_sales.to_sql('Fact_Sales', conn, if_exists='replace', index=False)

print("Data Warehouse created successfully!")

# Check tables inside the database
tables = pd.read_sql_query(
    "SELECT name FROM sqlite_master WHERE type='table';",
    conn
)

display(tables)
```

```python
# Step 8A: Prepare Customer Dimension for SCD Type 2

dim_customer_scd = dim_customer.copy()

dim_customer_scd['Effective_Start_Date'] = pd.to_datetime('2026-07-01')
dim_customer_scd['Effective_End_Date'] = pd.to_datetime('9999-12-31', errors='coerce')
dim_customer_scd['Is_Current'] = 1

print("CUSTOMER DIMENSION - FIRST PIPELINE EXECUTION")
display(dim_customer_scd)
```

```python
# Step 8B: Second Pipeline Execution - SCD Type 2

change_date = pd.Timestamp('2026-08-13')

# Close Kamal's old record
old_record = (
    (dim_customer_scd['Customer_ID'] == 'C001') &
    (dim_customer_scd['Is_Current'] == 1)
)

dim_customer_scd.loc[old_record, 'Effective_End_Date'] = change_date
dim_customer_scd.loc[old_record, 'Is_Current'] = 0

# Create new historical version
new_customer_version = pd.DataFrame({
    'Customer_SK': [dim_customer_scd['Customer_SK'].max() + 1],
    'Customer_ID': ['C001'],
    'Customer_Name': ['Kamal'],
    'City': ['Kandy'],
    'Effective_Start_Date': [change_date],
    'Effective_End_Date': [pd.NaT],
    'Is_Current': [1]
})

dim_customer_scd = pd.concat(
    [dim_customer_scd, new_customer_version],
    ignore_index=True
)

print("SCD TYPE 2 - SECOND PIPELINE EXECUTION")
display(
    dim_customer_scd[
        dim_customer_scd['Customer_ID'] == 'C001'
    ]
)
```

```python
# Step 9: Analytical Query - Sales by Product

product_sales = staging_sales.groupby(
    'Product_Name'
).agg(
    Total_Quantity=('Quantity', 'sum'),
    Total_Gross_Sales=('Gross_Sales', 'sum'),
    Total_Discount=('Discount', 'sum'),
    Total_Net_Sales=('Net_Sales', 'sum')
).reset_index()

product_sales = product_sales.sort_values(
    'Total_Net_Sales',
    ascending=False
)

print("SALES PERFORMANCE BY PRODUCT")
display(product_sales)
```

```python
# Step 10: Store Sales Analysis

store_sales = staging_sales.groupby(
    'Store_ID'
).agg(
    Number_of_Transactions=('Transaction_ID', 'count'),
    Total_Quantity=('Quantity', 'sum'),
    Total_Gross_Sales=('Gross_Sales', 'sum'),
    Total_Net_Sales=('Net_Sales', 'sum')
).reset_index()

store_sales = store_sales.merge(
    dim_store[['Store_ID', 'Store_Name']],
    on='Store_ID',
    how='left'
)

store_sales = store_sales[
    ['Store_ID', 'Store_Name', 'Number_of_Transactions',
     'Total_Quantity', 'Total_Gross_Sales', 'Total_Net_Sales']
]

print("SALES PERFORMANCE BY STORE")
display(store_sales)
```

```python
# Step 11: Customer Sales Analysis

customer_sales = staging_sales.groupby(
    ['Customer_ID', 'Customer_Name']
).agg(
    Number_of_Purchases=('Transaction_ID', 'count'),
    Total_Quantity=('Quantity', 'sum'),
    Total_Net_Sales=('Net_Sales', 'sum')
).reset_index()

customer_sales = customer_sales.sort_values(
    'Total_Net_Sales',
    ascending=False
)

print("CUSTOMER SALES ANALYSIS")
display(customer_sales)
```

```python
# Step 12: Create Second Source Data State

second_run_data = pd.DataFrame({
    'Transaction_ID': ['T011', 'T012', 'T013'],
    'Date': ['2026-08-13', '2026-08-14', '2026-08-15'],

    'Customer_ID': ['C001', 'C006', 'C002'],
    'Customer_Name': ['Kamal', 'Dilani', 'Nimali'],
    'City': ['Kandy', 'Gampaha', 'Kandy'],

    'Product_ID': ['P001', 'P004', 'P002'],
    'Product_Name': ['Laptop', 'Monitor', 'Mouse'],
    'Category': ['Electronics', 'Electronics', 'Accessories'],

    'Store_ID': ['S002', 'S001', 'S002'],

    'Quantity': [1, 1, 2],
    'Unit_Price': [250000, 65000, 3500],
    'Discount': [12000, 4000, 500]
})

second_run_data['Date'] = pd.to_datetime(second_run_data['Date'])

second_run_data['Gross_Sales'] = (
    second_run_data['Quantity'] * second_run_data['Unit_Price']
)

second_run_data['Net_Sales'] = (
    second_run_data['Gross_Sales'] - second_run_data['Discount']
)

print("SECOND SOURCE DATA STATE")
display(second_run_data)
```

```python
# Step 13: Process Second Pipeline Execution

# Add new customer Dilani to SCD Customer Dimension
new_customer = pd.DataFrame({
    'Customer_SK': [dim_customer_scd['Customer_SK'].max() + 1],
    'Customer_ID': ['C006'],
    'Customer_Name': ['Dilani'],
    'City': ['Gampaha'],
    'Effective_Start_Date': [pd.Timestamp('2026-08-14')],
    'Effective_End_Date': [pd.NaT],
    'Is_Current': [1]
})

dim_customer_scd = pd.concat(
    [dim_customer_scd, new_customer],
    ignore_index=True
)

# Get CURRENT customer surrogate keys
current_customers = dim_customer_scd[
    dim_customer_scd['Is_Current'] == 1
][['Customer_SK', 'Customer_ID']]

# Prepare second-run fact records
second_fact = second_run_data.merge(
    current_customers,
    on='Customer_ID',
    how='left'
)

second_fact = second_fact.merge(
    dim_product[['Product_SK', 'Product_ID']],
    on='Product_ID',
    how='left'
)

second_fact = second_fact.merge(
    dim_store[['Store_SK', 'Store_ID']],
    on='Store_ID',
    how='left'
)

print("SECOND PIPELINE - CUSTOMER DIMENSION")
display(dim_customer_scd)

print("\nSECOND PIPELINE - NEW FACT RECORDS")
display(
    second_fact[
        ['Transaction_ID', 'Customer_SK', 'Product_SK',
         'Store_SK', 'Quantity', 'Gross_Sales', 'Net_Sales']
    ]
)
```

```python
# Step 14: Update Fact Table After Second Pipeline Execution

# Add simple Date_SK values for the new August dates
second_fact['Date_SK'] = [11, 12, 13]

second_fact_final = second_fact[
    [
        'Transaction_ID',
        'Date_SK',
        'Customer_SK',
        'Product_SK',
        'Store_SK',
        'Quantity',
        'Unit_Price',
        'Discount',
        'Gross_Sales',
        'Net_Sales'
    ]
]

# Combine first and second pipeline fact data
updated_fact_sales = pd.concat(
    [fact_sales, second_fact_final],
    ignore_index=True
)

print("UPDATED FACT SALES TABLE AFTER SECOND PIPELINE EXECUTION")
display(updated_fact_sales)
```

```python
# Step 15: Final Data Warehouse Validation

print("========== FINAL DATA WAREHOUSE VALIDATION ==========")

print("\nTotal Fact Records:")
print(len(updated_fact_sales))

print("\nTotal Customer Dimension Records:")
print(len(dim_customer_scd))

print("\nCurrent Customers:")
print(len(dim_customer_scd[dim_customer_scd['Is_Current'] == 1]))

print("\nHistorical Customer Records:")
print(len(dim_customer_scd[dim_customer_scd['Is_Current'] == 0]))

print("\nTotal Net Sales:")
print(updated_fact_sales['Net_Sales'].sum())

print("\nDuplicate Transaction IDs:")
print(updated_fact_sales['Transaction_ID'].duplicated().sum())

print("\nMissing Customer Keys:")
print(updated_fact_sales['Customer_SK'].isnull().sum())

print("\nVALIDATION COMPLETED SUCCESSFULLY")
```
